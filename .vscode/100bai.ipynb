```python
#bai6
import math
a, b, c = map(int, input().split())
cv=a+b+c
p=cv/2
s=(p*(p-a)*(p-b)*(p-c))
print(f"{cv:.1f}", f"{(math.sqrt(s)):.3f}")
```
