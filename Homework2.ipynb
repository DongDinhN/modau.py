```python
#Bài 1: Nhập vào 2 cạnh của một hình chữ nhật. In ra màn hình diệntíchvàchu vi của nó. 
a, b = map(int, input().split())
dt = a*b
cv = 2*(a + b)
print(dt, cv)
```

```python
#Bài 2: Nhập vào bán kính của hình tròn. In ra màn hình diện tích và chu vi của nó. 
r = int(input())
dt = 3.14*r**2
cv = 2*3.14*r
print(f"{dt:.2f}", f"{cv:.2f}")
```

```python
#Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.
import math
a, b, c = map(int, input().split())
cv = a+b+c
p = (a+b+c)/2
s = (math.sqrt(p*(p-a)*(p-b)*(p-c)))
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        Loai_tam_giac = "Tam giac deu"
    elif a==b or a==c or b==c:
        Loai_tam_giac = "tam giac can"
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
        Loai_tam_giac = "Tam giac vuong"
    else:
        Loai_tam_giac = "Tam giac thuong"
    print(Loai_tam_giac)
    print(f"Chu vi va dien tich {cv:.2f}, {s:.2f}")
else:
    print("Khong la ba canh cua mot tam giac")

    

    
```

```python
#Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)
import math
a, b, c = map(int, input().split())
delta = b**2-4*a*c
if a == 0:
    if b == 0 and c != 0:
        print("Phuong trinh vo mghiem")
    elif b==0 and c ==0:
        print("Phuong trinh co vo so nghiem")
    elif b != 0 and c != 0:
        print(f"Phuong trinh co mot nghiem la {-c/b:.2f}")
if a != 0:
    if delta > 0:
        x1 = (-b + math.sqrt(delta))/(2*a)
        x2 = (-b - math.sqrt(delta))/(2*a)
        print(f"Phuong trinh co 2 nghiem phan biet la: {x1, x2:.2f}")
    elif delta == 0:
        print(f"Phuong trinh co nghiem kep: {x1:.2f}")
    elif delta < 0:
        print("Phuong trinh vo nghiem")

```

```python
#Bài 8: Giải hệ phương trình tuyến tính: ax + by = m
#                                        cx + dy = n
a, b, m = map(int, input().split())
c, d, n = map(int, input().split())
D = a*d-c*b                              #Dinh thuc chinh
Dx = m*d-b*n                             #Dinh thuc theo x
Dy = a*n-c*m                             #Dinh thuc theo y
if D != 0:
    print(f"He phuong trinh co nghiem duy nhat x = {Dx/D:.2f} y = {Dy/D:.2f}")
elif Dx == 0 and Dy == 0:
    print("He phuong trinh co vo so nghiem")
else:
    print("Phuong trinh vo nghiem")
```
